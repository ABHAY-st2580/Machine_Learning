```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score
```

```python
X, y = load_iris(return_X_y = True)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
lr = LogisticRegression()
dt = DecisionTreeClassifier(max_depth = 20)
svc = SVC()
```

```python
estimators = [('lr', lr), ('dt', dt), ('svc', svc)]
```

```python
for estimator in estimators:
    scores = cross_val_score(estimator[1], X, y, cv = 10)
    print(estimator[0],':', np.round(np.mean(scores), 4)*100)
```

```python
from sklearn.ensemble import VotingClassifier, VotingRegressor
```

```python
vc = VotingClassifier(estimators)
```

```python
scores = cross_val_score(vc, X, y, cv = 30)
print('Voting_Classifer:', np.round(np.mean(scores), 7)*100)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.svm import SVC

# Generate sample dataset (2 features)
X, y = make_classification(n_samples=200,
                           n_features=2,
                           n_redundant=0,
                           n_clusters_per_class=1,
                           random_state=42)

model = SVC(kernel='poly')
model.fit(X, y)

x_min, x_max = X[:,0].min()-1, X[:,0].max()+1
y_min, y_max = X[:,1].min()-1, X[:,1].max()+1

xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 500),
    np.linspace(y_min, y_max, 500)
)

Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contourf(xx, yy, Z, alpha=0.3)

plt.scatter(X[:,0], X[:,1], c=y)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Decision Boundary")
plt.show()
```
